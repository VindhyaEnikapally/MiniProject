```python
def what_if_prediction(student_data, changes):
    hypothetical_student = student_data.copy()

    for feature, value in changes.items():
        hypothetical_student[feature] = value

    # Convert Series to DataFrame
    hypothetical_df = hypothetical_student.to_frame().T

    # Apply the same preprocessing
    hypothetical_processed = preprocessor.transform(hypothetical_df)

    # Predict risk probability
    risk_probability = best_xgb.predict_proba(hypothetical_processed)[0, 1]

    return risk_probability
```

```python
def intervention_policy(student, risk_probability):
    
    interventions = []

    # Attendance
    if student['attendance_percentage'] < 75:
        interventions.append({
            'priority': 'High',
            'factor': 'Attendance',
            'action': 'Attendance improvement plan and faculty follow-up'
        })
    elif student['attendance_percentage'] < 85:
        interventions.append({
            'priority': 'Medium',
            'factor': 'Attendance',
            'action': 'Monitor attendance and encourage regular class participation'
        })

    # Study hours
    if student['study_hours_per_day'] < 3:
        interventions.append({
            'priority': 'High',
            'factor': 'Study Hours',
            'action': 'Create a structured daily study schedule'
        })
    elif student['study_hours_per_day'] < 4:
        interventions.append({
            'priority': 'Medium',
            'factor': 'Study Hours',
            'action': 'Encourage additional daily study time'
        })

    # Practice tests
    if student['practice_tests_completed'] < 2:
        interventions.append({
            'priority': 'High',
            'factor': 'Practice Tests',
            'action': 'Assign additional practice tests and review sessions'
        })
    elif student['practice_tests_completed'] < 4:
        interventions.append({
            'priority': 'Medium',
            'factor': 'Practice Tests',
            'action': 'Encourage regular practice-test completion'
        })

    # Preparation days
    if student['exam_preparation_days'] < 7:
        interventions.append({
            'priority': 'High',
            'factor': 'Exam Preparation',
            'action': 'Create an early exam-preparation plan'
        })

    # Stress
    if student['stress_level'] >= 7:
        interventions.append({
            'priority': 'High',
            'factor': 'Stress',
            'action': 'Faculty mentoring and stress-management support'
        })

    # Motivation
    if student['motivation_level'] == 'Low':
        interventions.append({
            'priority': 'Medium',
            'factor': 'Motivation',
            'action': 'One-to-one mentoring and goal-setting'
        })

    # Time management
    if student['time_management_score'] < 60:
        interventions.append({
            'priority': 'Medium',
            'factor': 'Time Management',
            'action': 'Provide a weekly study and task-planning schedule'
        })

    # Overall risk
    if risk_probability >= 0.70:
        monitoring = 'Weekly faculty monitoring'
    elif risk_probability >= 0.40:
        monitoring = 'Biweekly faculty monitoring'
    else:
        monitoring = 'Routine monitoring'

    return interventions, monitoring
```

```python
def shap_intervention_policy(
    student,
    shap_values_student,
    feature_names,
    risk_probability
):
    
    interventions = []

    intervention_map = {
        'practice_tests_completed': {
            'shap_feature': 'num__practice_tests_completed',
            'factor': 'Practice Tests',
            'condition': lambda x: x < 4,
            'action': 'Encourage regular practice-test completion'
        },

        'exam_preparation_days': {
            'shap_feature': 'num__exam_preparation_days',
            'factor': 'Exam Preparation',
            'condition': lambda x: x < 7,
            'action': 'Create an early exam-preparation plan'
        },

        'attendance_percentage': {
            'shap_feature': 'num__attendance_percentage',
            'factor': 'Attendance',
            'condition': lambda x: x < 85,
            'action': 'Monitor attendance and create an attendance improvement plan'
        },

        'study_hours_per_day': {
            'shap_feature': 'num__study_hours_per_day',
            'factor': 'Study Hours',
            'condition': lambda x: x < 4,
            'action': 'Encourage additional structured study time'
        },

        'stress_level': {
            'shap_feature': 'num__stress_level',
            'factor': 'Stress',
            'condition': lambda x: x >= 7,
            'action': 'Faculty mentoring and stress-management support'
        },

        'exam_anxiety_level': {
            'shap_feature': 'num__exam_anxiety_level',
            'factor': 'Exam Anxiety',
            'condition': lambda x: x >= 7,
            'action': 'Provide exam-anxiety management and mentoring support'
        },

        'time_management_score': {
            'shap_feature': 'num__time_management_score',
            'factor': 'Time Management',
            'condition': lambda x: x < 60,
            'action': 'Provide a structured weekly planning schedule'
        }
    }

    # SHAP feature → SHAP value
    shap_lookup = dict(zip(feature_names, shap_values_student))

    for raw_feature, rule in intervention_map.items():

        if raw_feature in student.index:

            actual_value = student[raw_feature]

            shap_value = shap_lookup.get(
                rule['shap_feature'],
                0
            )

            # Only recommend if the student's value
            # meets the condition AND SHAP pushes risk upward
            if rule['condition'](actual_value) and shap_value > 0:

                if shap_value >= 0.30:
                    priority = 'High'
                elif shap_value >= 0.10:
                    priority = 'Medium'
                else:
                    priority = 'Low'

                interventions.append({
                    'Priority': priority,
                    'Factor': rule['factor'],
                    'SHAP Contribution': round(float(shap_value), 3),
                    'Current Value': actual_value,
                    'Action': rule['action']
                })

    # Sort by SHAP contribution
    interventions = sorted(
        interventions,
        key=lambda x: x['SHAP Contribution'],
        reverse=True
    )

    return interventions
```

```python
def final_intervention_engine(
    student,
    shap_values_student,
    feature_names,
    risk_probability
):

    interventions = []

    # Actionable numerical factors
    rules = {
        'practice_tests_completed': {
            'shap_feature': 'num__practice_tests_completed',
            'condition': lambda x: x < 4,
            'factor': 'Practice Tests',
            'action': 'Encourage regular practice-test completion'
        },

        'exam_preparation_days': {
            'shap_feature': 'num__exam_preparation_days',
            'condition': lambda x: x < 7,
            'factor': 'Exam Preparation',
            'action': 'Create an early exam-preparation plan'
        },

        'attendance_percentage': {
            'shap_feature': 'num__attendance_percentage',
            'condition': lambda x: x < 85,
            'factor': 'Attendance',
            'action': 'Monitor attendance and provide an attendance improvement plan'
        },

        'study_hours_per_day': {
            'shap_feature': 'num__study_hours_per_day',
            'condition': lambda x: x < 4,
            'factor': 'Study Hours',
            'action': 'Encourage structured additional study time'
        },

        'sleep_hours': {
            'shap_feature': 'num__sleep_hours',
            'condition': lambda x: x < 6,
            'factor': 'Sleep',
            'action': 'Encourage healthier sleep and study scheduling'
        },

        'stress_level': {
            'shap_feature': 'num__stress_level',
            'condition': lambda x: x >= 7,
            'factor': 'Stress',
            'action': 'Faculty mentoring and stress-management support'
        },

        'time_management_score': {
            'shap_feature': 'num__time_management_score',
            'condition': lambda x: x < 60,
            'factor': 'Time Management',
            'action': 'Provide structured weekly planning support'
        },

        'exam_anxiety_level': {
            'shap_feature': 'num__exam_anxiety_level',
            'condition': lambda x: x >= 7,
            'factor': 'Exam Anxiety',
            'action': 'Provide exam-anxiety management and mentoring support'
        },

        'self_study_hours': {
            'shap_feature': 'num__self_study_hours',
            'condition': lambda x: x < 2,
            'factor': 'Self Study',
            'action': 'Encourage consistent self-study sessions'
        },

        'assignment_completion_rate': {
            'shap_feature': 'num__assignment_completion_rate',
            'condition': lambda x: x < 70,
            'factor': 'Assignment Completion',
            'action': 'Monitor assignment completion and provide academic follow-up'
        }
    }

    shap_lookup = dict(zip(feature_names, shap_values_student))

    for raw_feature, rule in rules.items():

        actual_value = student[raw_feature]

        shap_value = float(
            shap_lookup.get(rule['shap_feature'], 0)
        )

        # Only recommend when:
        # student's value needs attention AND
        # the model pushes risk upward
        if rule['condition'](actual_value) and shap_value > 0:

            if shap_value >= 0.30:
                priority = 'High'
            elif shap_value >= 0.10:
                priority = 'Medium'
            else:
                priority = 'Low'

            interventions.append({
                'Priority': priority,
                'Factor': rule['factor'],
                'SHAP Contribution': round(shap_value, 3),
                'Current Value': actual_value,
                'Action': rule['action']
            })

    interventions.sort(
        key=lambda x: x['SHAP Contribution'],
        reverse=True
    )

    return interventions
```

```python
def final_shap_intervention_policy(
    student,
    shap_values_student,
    feature_names,
    risk_probability
):
    
    intervention_map = {
        'practice_tests_completed': {
            'shap_feature': 'num__practice_tests_completed',
            'condition': lambda x: x < 4,
            'action': 'Encourage regular practice-test completion'
        },

        'exam_preparation_days': {
            'shap_feature': 'num__exam_preparation_days',
            'condition': lambda x: x < 7,
            'action': 'Create an early exam-preparation plan'
        },

        'stress_level': {
            'shap_feature': 'num__stress_level',
            'condition': lambda x: x >= 7,
            'action': 'Faculty mentoring and stress-management support'
        },

        'exam_anxiety_level': {
            'shap_feature': 'num__exam_anxiety_level',
            'condition': lambda x: x >= 7,
            'action': 'Provide exam-anxiety management and mentoring support'
        },

        'attendance_percentage': {
            'shap_feature': 'num__attendance_percentage',
            'condition': lambda x: x < 85,
            'action': 'Monitor attendance and encourage regular class participation'
        },

        'study_hours_per_day': {
            'shap_feature': 'num__study_hours_per_day',
            'condition': lambda x: x < 4,
            'action': 'Encourage a consistent daily study schedule'
        },

        'sleep_hours': {
            'shap_feature': 'num__sleep_hours',
            'condition': lambda x: x < 6,
            'action': 'Encourage healthier sleep habits'
        },

        'time_management_score': {
            'shap_feature': 'num__time_management_score',
            'condition': lambda x: x < 60,
            'action': 'Provide time-management guidance'
        },

        'self_study_hours': {
            'shap_feature': 'num__self_study_hours',
            'condition': lambda x: x < 2,
            'action': 'Encourage additional self-study time'
        },

        'assignment_completion_rate': {
            'shap_feature': 'num__assignment_completion_rate',
            'condition': lambda x: x < 70,
            'action': 'Monitor assignment completion and provide academic support'
        }
    }

    shap_lookup = dict(zip(feature_names, shap_values_student))

    interventions = []

    for raw_feature, rule in intervention_map.items():

        actual_value = student[raw_feature]
        shap_value = shap_lookup.get(rule['shap_feature'], 0)

        # Intervention only when:
        # 1. The raw value meets the intervention condition
        # 2. SHAP shows that the factor increases this student's risk
        if rule['condition'](actual_value) and shap_value > 0:

            if shap_value >= 0.30:
                priority = 'High'
            elif shap_value >= 0.10:
                priority = 'Medium'
            else:
                priority = 'Low'

            interventions.append({
                'Priority': priority,
                'Factor': raw_feature,
                'SHAP Contribution': round(float(shap_value), 3),
                'Current Value': actual_value,
                'Action': rule['action']
            })

    priority_order = {
        'High': 0,
        'Medium': 1,
        'Low': 2
    }

    interventions.sort(
        key=lambda x: priority_order[x['Priority']]
    )

    return pd.DataFrame(interventions)
```
